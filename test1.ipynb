```python
import ast 
def convert(obj):
    L=[]
    for i in ast.literal_eval(obj):
        L.append(i['name'])
    return L
```

```python
moviesf['genres'] = moviesf['genres'].apply(convert)
```

```python
moviesf.head()
```

```python
moviesf['keywords'] = moviesf['keywords'].apply(convert)
```

```python
moviesf.head(3)
```

```python
def convert3(obj):
    L=[]
    counter=0
    for i in ast.literal_eval(obj):
        if counter !=3:
            L.append(i['name'])
            counter +=1
        else:
            break
    return L
```

```python
moviesf['cast'] = moviesf['cast'].apply(convert3)    
```

```python

def fetch_director(obj):
    L=[]
    for i in ast.literal_eval(obj):
        if i['job']=='Director':
            L.append(i['name'])
            break
    return L
```

```python
moviesf['crew']=moviesf['crew'].apply(fetch_director)
```

```python
moviesf.head(10)
```

```python
moviesf['overview']=moviesf['overview'].apply(lambda x:x.split())
```

```python
moviesf.head(2)
```

```python
moviesf['genres']=moviesf['genres'].apply(lambda x:[i.replace(" ","") for i in x])
moviesf['keywords']=moviesf['keywords'].apply(lambda x:[i.replace(" ","") for i in x])
moviesf['cast']=moviesf['cast'].apply(lambda x:[i.replace(" ","") for i in x])
moviesf['crew']=moviesf['crew'].apply(lambda x:[i.replace(" ","") for i in x])
```

```python
moviesf.head(2)
```

```python
moviesf['tags']=moviesf['overview']+moviesf['genres']+moviesf['keywords']+moviesf['cast']+moviesf['crew']
```

```python
moviesf.head(2)
```

```python
new=moviesf[['movie_id','title','tags']]
new.head(2)
```

```python
new['tags']=new['tags'].apply(lambda x:" ".join(x))
```

```python
new['tags'][0]
```

```python
new['tags']=new['tags'].apply(lambda x:x.lower())
```

```python
new.head(3)
```

```python
new['tags'][0]
```

```python
import nltk 
```

```python
from nltk.stem.porter import PorterStemmer
ps=PorterStemmer()
```

```python
def stem(text):
    y=[]
    for i in text.split():
        y.append(ps.stem(i))

    return " ".join(y)

```

```python
new['tags']=new['tags'].apply(stem)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv=CountVectorizer(max_features=5000,stop_words='english')
vectors=cv.fit_transform(new['tags']).toarray()
```

```python
vectors
```

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity = cosine_similarity(vectors)
```

```python
sorted(list(enumerate(similarity[0])),reverse=True,key=lambda x:x[1])[1:6]
```

```python
similarity[1]
```

```python
def recommend(movie):
    movie_index=new[new['title']==movie].index[0]
    distances=similarity[movie_index]
    movies_list=sorted(list(enumerate(distances)),reverse=True,key=lambda x:x[1])[1:6]

    for i in movies_list:
        print(new.iloc[i[0]].title)
```

```python
recommend('Avatar')
```

```python
recommend('Hitman')
```
